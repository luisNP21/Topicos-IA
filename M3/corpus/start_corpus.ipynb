## 5. Verificar resultado

```python
# Celda de verificacion — confirma que todas las fuentes de fuentes.yaml quedaron indexadas

import json
from pathlib import Path
from corpus_store import cargar_coleccion
from ingesta import cargar_fuentes
import sys

sys.path.append('/content/corpus/M3/corpus')

ROOT = Path("/content/corpus/M3/corpus")

# 1. Fuentes esperadas segun fuentes.yaml
fuentes = cargar_fuentes(f"{ROOT}/fuentes.yaml")
doc_ids_esperados = {f["doc_id"] for f in fuentes}

# 2. Fuentes que quedaron en el manifest
manifest_path = Path(PROJECT_ROOT) / "data/guias_clinicas/corpus_manifest.json"
with open(manifest_path, "r", encoding="utf-8") as f:
    manifest = json.load(f)
doc_ids_manifest = {m["doc_id"] for m in manifest}

# 3. Fuentes que realmente tienen chunks en disco
chunks_dir = Path(PROJECT_ROOT) / "data/guias_clinicas/chunks"
doc_ids_con_chunks = {
    p.stem.rsplit("_chunk", 1)[0] for p in chunks_dir.glob("*.json")
}

# 4. Fuentes que están indexadas en Chroma (via metadata doc_id)
coleccion = cargar_coleccion(persist_dir=f"{PROJECT_ROOT}/data/chroma_guias")
todo = coleccion.get(include=["metadatas"])
doc_ids_chroma = {m["doc_id"] for m in todo["metadatas"]}

# --- Reporte ---
print(f"Fuentes en fuentes.yaml: {len(doc_ids_esperados)}")
print(f"Fuentes en manifest:     {len(doc_ids_manifest)}")
print(f"Fuentes con chunks JSON: {len(doc_ids_con_chunks)}")
print(f"Fuentes en Chroma:       {len(doc_ids_chroma)}")
print(f"Total chunks en Chroma:  {coleccion.count()}\n")

faltantes_manifest = doc_ids_esperados - doc_ids_manifest
faltantes_chunks = doc_ids_esperados - doc_ids_con_chunks
faltantes_chroma = doc_ids_esperados - doc_ids_chroma

if not (faltantes_manifest or faltantes_chunks or faltantes_chroma):
    print("OK: todas las fuentes de fuentes.yaml están presentes en manifest, chunks y Chroma.")
else:
    if faltantes_manifest:
        print(f"FALTAN en manifest: {faltantes_manifest}")
    if faltantes_chunks:
        print(f"FALTAN chunks JSON: {faltantes_chunks}")
    if faltantes_chroma:
        print(f"FALTAN en Chroma: {faltantes_chroma}")
```

### 5.1. Herramienta extra: Eliminar guía

```python
import json
from pathlib import Path
from corpus_store import cargar_coleccion


def eliminar_guia(doc_id: str, project_root: str, chunks_dir: str, chroma_dir: str,
                  manifest_path: str, pdf_relpath: str = None, dry_run: bool = True) -> dict:
    """
    Elimina doc_id de: chunks JSON, coleccion Chroma y manifest.
    Opcionalmente borra el PDF fuente (pdf_relpath, relativo a project_root).
    dry_run=True solo reporta lo que borraria, sin tocar nada.
    """
    root = Path(project_root)
    chunks_path = root / chunks_dir
    manifest_file = root / manifest_path

    # 1. Chunks JSON
    archivos = sorted(chunks_path.glob(f"{doc_id}_chunk*.json"))

    # 2. Chroma
    coleccion = cargar_coleccion(persist_dir=str(root / chroma_dir))
    ids_chroma = coleccion.get(where={"doc_id": doc_id}, include=[])["ids"]

    # 3. Manifest
    manifest = []
    if manifest_file.exists():
        with open(manifest_file, "r", encoding="utf-8") as f:
            manifest = json.load(f)
    en_manifest = any(m["doc_id"] == doc_id for m in manifest)

    reporte = {
        "doc_id": doc_id,
        "chunks_json": len(archivos),
        "puntos_chroma": len(ids_chroma),
        "en_manifest": en_manifest,
        "pdf": pdf_relpath,
        "dry_run": dry_run,
    }
    print(reporte)

    if dry_run:
        print("DRY RUN: no se borro nada. Repite con dry_run=False para aplicar.")
        return reporte

    for a in archivos:
        a.unlink()
    if ids_chroma:
        coleccion.delete(ids=ids_chroma)
    if en_manifest:
        manifest = [m for m in manifest if m["doc_id"] != doc_id]
        with open(manifest_file, "w", encoding="utf-8") as f:
            json.dump(manifest, f, ensure_ascii=False, indent=2)
    if pdf_relpath:
        pdf = root / pdf_relpath
        if pdf.exists():
            pdf.unlink()

    print(f"Eliminado {doc_id}. Chunks restantes en Chroma: {coleccion.count()}")
    return reporte
```

### 5.2. Herramienta extra: Verificaciones

```python
# Celda: distribucion de tokens por chunk (Chroma no guarda el conteo, se calcula con el tokenizer)
import pandas as pd
from corpus_store import cargar_coleccion
from config_utils import cargar_config
from chunking import cargar_tokenizer

config = cargar_config(f"{ROOT}/config.yaml")
model_name = config["embeddings"]["model_name"]
tokenizer = cargar_tokenizer(model_name)

coleccion = cargar_coleccion(persist_dir=f"{PROJECT_ROOT}/data/chroma_guias")
todo = coleccion.get(include=["documents", "metadatas"])

# Se cuenta el texto tal como lo ve el modelo al indexar: prefijo "passage: " + tokens especiales
textos = [f"passage: {t}" for t in todo["documents"]]
n_tokens = [len(ids) for ids in tokenizer(textos, add_special_tokens=True)["input_ids"]]

df = pd.DataFrame({
    "chunk_id": todo["ids"],
    "doc_id": [m["doc_id"] for m in todo["metadatas"]],
    "categoria": [m.get("categoria_seccion", "") for m in todo["metadatas"]],
    "n_tokens": n_tokens,
})

max_tokens_cfg = config["chunking"]["max_tokens"]
limite_modelo = tokenizer.model_max_length  # 512 en multilingual-e5-base

print(f"Chunks: {len(df)} | max_tokens en config: {max_tokens_cfg} | limite del modelo: {limite_modelo}\n")
print(df["n_tokens"].describe(percentiles=[0.5, 0.9, 0.99]).round(1), "\n")
print(f"Chunks sobre max_tokens del config ({max_tokens_cfg}): {(df['n_tokens'] > max_tokens_cfg).sum()}")
print(f"Chunks sobre el limite del modelo ({limite_modelo}, se truncan al embeber): {(df['n_tokens'] > limite_modelo).sum()}\n")

print("Por documento:")
print(df.groupby("doc_id")["n_tokens"].agg(["count", "min", "mean", "max"]).round(1), "\n")
print("Por categoria:")
print(df.groupby("categoria")["n_tokens"].agg(["count", "min", "mean", "max"]).round(1))
```

```python
# Cuenta chunks consecutivos de la misma seccion donde el siguiente es sufijo del anterior
ids = todo["ids"]; docs = todo["documents"]; secs = [m["seccion"] for m in todo["metadatas"]]
orden = sorted(range(len(ids)), key=lambda i: (todo["metadatas"][i]["doc_id"], int(ids[i].rsplit("_chunk", 1)[1])))
sufijos = sum(1 for a, b in zip(orden, orden[1:]) if secs[a] == secs[b] and docs[a].endswith(docs[b]))
print(f"Chunks que son sufijo del anterior: {sufijos} de {len(ids)}")
```
