## 5. Verificar resultado

```python
# Celda de verificacion — confirma que todas las fuentes de fuentes.yaml quedaron indexadas

import json
from pathlib import Path
from corpus_store import cargar_coleccion
from ingesta import cargar_fuentes
import sys

sys.path.append('/content/corpus/M3/corpus')

ROOT = Path("/content/corpus/M3/corpus")

# 1. Fuentes esperadas segun fuentes.yaml
fuentes = cargar_fuentes(f"{ROOT}/fuentes.yaml")
doc_ids_esperados = {f["doc_id"] for f in fuentes}

# 2. Fuentes que quedaron en el manifest
manifest_path = Path(PROJECT_ROOT) / "data/guias_clinicas/corpus_manifest.json"
with open(manifest_path, "r", encoding="utf-8") as f:
    manifest = json.load(f)
doc_ids_manifest = {m["doc_id"] for m in manifest}

# 3. Fuentes que realmente tienen chunks en disco
chunks_dir = Path(PROJECT_ROOT) / "data/guias_clinicas/chunks"
doc_ids_con_chunks = {
    p.stem.rsplit("_chunk", 1)[0] for p in chunks_dir.glob("*.json")
}

# 4. Fuentes que están indexadas en Chroma (via metadata doc_id)
coleccion = cargar_coleccion(persist_dir=f"{PROJECT_ROOT}/data/chroma_guias")
todo = coleccion.get(include=["metadatas"])
doc_ids_chroma = {m["doc_id"] for m in todo["metadatas"]}

# --- Reporte ---
print(f"Fuentes en fuentes.yaml: {len(doc_ids_esperados)}")
print(f"Fuentes en manifest:     {len(doc_ids_manifest)}")
print(f"Fuentes con chunks JSON: {len(doc_ids_con_chunks)}")
print(f"Fuentes en Chroma:       {len(doc_ids_chroma)}")
print(f"Total chunks en Chroma:  {coleccion.count()}\n")

faltantes_manifest = doc_ids_esperados - doc_ids_manifest
faltantes_chunks = doc_ids_esperados - doc_ids_con_chunks
faltantes_chroma = doc_ids_esperados - doc_ids_chroma

if not (faltantes_manifest or faltantes_chunks or faltantes_chroma):
    print("OK: todas las fuentes de fuentes.yaml están presentes en manifest, chunks y Chroma.")
else:
    if faltantes_manifest:
        print(f"FALTAN en manifest: {faltantes_manifest}")
    if faltantes_chunks:
        print(f"FALTAN chunks JSON: {faltantes_chunks}")
    if faltantes_chroma:
        print(f"FALTAN en Chroma: {faltantes_chroma}")
```
